```python
var('x,y,z,w')

f1 = 1 - (x^4 + y^4 + z^4 + w^4);
f2 = 1 - ((x-1)^4 + y^4 + z^4 + w^4);
#f3 = 1 - ((x+1/2)^4 + (y-1)^4);
#f4 = 1 - ((x+2/3)^4 + (y-6/5)^4);

#implicit_plot(f1*f2*f3*f4==0, (x,-2,3), (y,-2,2))
```

```python
var('t')
Rt.<x,y,z,w,t> = PolynomialRing(ZZ)   
Ft = Rt.fraction_field()
Wt.<Dx,Dy,Dz,Dw,Dt> = OreAlgebra(Ft)

ft = Rt(f1*f2) - t; ft
```

```python
At = Ft(y*ft.derivative(y)/ft)
At
```

```python
annt = Wt.ideal([At*D-D(At) for D in Wt.gens()]); annt
```

```python
ct1 = annt.ct(Dy, certificates=False)
```

```python
ct2 = ct1[0].parent().ideal(ct1).ct(Dw, certificates=False)
```

```python
ct3 = ct2[0].parent().ideal(ct2).ct(Dz, certificates=False)
```

```python
ct4 = ct3[0].parent().ideal(ct3).ct(Dx, certificates=False); ct4
```

```python
ct4[0].order()
```

```python
maxt = QQ[t](ct4[0].leading_coefficient()).roots(QQbar, multiplicities=False)
```

```python
maxt[4]
```
