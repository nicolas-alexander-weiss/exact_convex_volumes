### Example of convex body with non-convex deformation:
The following is an example of a convex body, given by 
```math
C = \{f>0\} \cap \{g_1>0\} \cap \{g_2>0\} \{g_3>0\}
```
However, the deformation (hour-glass figure, in grey)
```math
C_t = C \cap \{f\cdot g_1 \cdot g_2 \cdot g_3 - t > 0\}
```
is not convex.

````python
var('x y')

eps = 1/2
f = -(y^2-(x+1)*x^2 -eps)
g1 = -(x - 2)
g2 = (y + eps)
g3 = -(y - eps)

contour_plot(f*g1*g2*g3, (x,-2,2), (y,-1,1), fill=False, plot_points=150, contours=[0, 0.2])
````
