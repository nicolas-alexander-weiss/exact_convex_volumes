### Example Set: Intersections of Lp balls in RR2

```sage
S = PolynomialRing(QQ, "x", 2)

prec_bits = 400
computed_volumes = {}
```

```sage
# Two L2 balls - Ex 1:

p = 2
p_mus = ((p,(0,0)), (p,(1,0)))
fs = [shifted_lp_poly(S, p, mu) for p,mu in p_mus]

computed_volumes[p_mus] = volume2(fs,prec=prec_bits, debug_level=1)

print(f"\nVolume of the intersection of two L2 balls with centers {[mu for p,mu in p_mus]} equals = {computed_volumes[p_mus]}")
```

```sage
# Two L2 balls - Ex 2:

p = 2
p_mus = ((p,(0,0)), (p,(0,1)))
fs = [shifted_lp_poly(S, p, mu) for p,mu in p_mus]

computed_volumes[p_mus] = volume2(fs,prec=prec_bits, debug_level=0)

print(f"\nVolume of the intersection of two L2 balls with centers {[mu for p,mu in p_mus]} equals = {computed_volumes[p_mus]}")
```

```sage
# Two L2 balls - Ex 3:

p = 2
p_mus = ((p,(1/2,0)), (p,(1/2,1)))
fs = [shifted_lp_poly(S, p, mu) for p,mu in p_mus]

computed_volumes[p_mus] = volume2(fs,prec=prec_bits, debug_level=0)

print(f"\nVolume of the intersection of two L2 balls with centers {[mu for p,mu in p_mus]} equals = {computed_volumes[p_mus]}")
```

Let's try another translation.

```sage
# Two L2 balls - Ex 4:


p = 2
p_mus = ((p,(1/2,1/3)), (p,(1/2,4/3)))
fs = [shifted_lp_poly(S, p, mu) for p,mu in p_mus]

computed_volumes[p_mus] = volume2(fs,prec=prec_bits, debug_level=0)

print(f"\nVolume of the intersection of two L2 balls with centers {[mu for p,mu in p_mus]} equals = {computed_volumes[p_mus]}")
```

Let us now compute the volume for the intersection of two l2 balls with centers (0,0) and (1,1). This computation takes significantly longer.

```sage
# Two L2 balls - Ex 5:

p = 2
p_mus = ((p,(0,0)), (p,(1,1)))
fs = [shifted_lp_poly(S, p, mu) for p,mu in p_mus]

computed_volumes[p_mus] = volume2(fs,prec=prec_bits, debug_level=0)

print(f"\nVolume of the intersection of two L2 balls with centers {[mu for p,mu in p_mus]} equals = {computed_volumes[p_mus]}")
```
