## Four dimensional example - intersection of two Euclidean balls 

```python
def example_translation_RR4():
    n = 4
    p = 2

    S = PolynomialRing(QQ, "x", n)
    x = S.gens()

    mus = [[0,0,0,0], [1,0,0,0]]

    fs = [shifted_lp_poly(S, p, mu) for mu in mus]

    print("Computing the volume of the intersection of two l2 balls in RR4 centered at {}".format(mus))
    prec = 1000
    strategy={"proj_var":last_variable_proj_var_strategy}
    vol = volume2(fs, prec, strategy=strategy, debug_level=3)

    print("The volume = {}".format(vol))
```

## Two $\ell_4$ balls in 2 dimensions
Centered at $(0,0)$ and $(a,0)$. We vary $a$ to get different Picard-Fuchs operators.

```python
#Picard Fuchs operator in t for 2 l4 balls centered at (0,0) and (a,0)
def special_pointsl42():
    n = 2
    S = PolynomialRing(QQ, "x", n)
    x = S.gens()
    A = [-2, 0, 2]
    L = []
    for a in A:
        mus = [[0,0], [a,0]]

        fs = [shifted_lp_poly(S, 4, mu) for mu in mus]

        print("Computing the Picard-Fuchs of the intersection of two l2 balls in RR2 centered at {}".format(mus))
        #prec = 200
        strategy={"proj_var":last_variable_proj_var_strategy}
        L.append(get_picard_fuchs_t(fs, strategy=strategy, debug_level=1))

    print("The special Picard-Fuchs operators are = {}".format(L))

def generic_pointsl42(a):
    n = 2
    S = PolynomialRing(QQ, "x", n)
    x = S.gens()
    mus = [[0,0], [-2,0]]

    fs = [shifted_lp_poly(S, 4, mu) for mu in mus]

    print("Computing the volume of the Picard-Fuchs of two l2 balls in RR2 centered at {}".format(mus))
    #prec = 200
    strategy={"proj_var":last_variable_proj_var_strategy}
    P = get_picard_fuchs_t(fs, strategy=strategy, debug_level=1)

    print("The special Picard-Fuchs operator is = {}".format(P))
```

## Two Euclidean balls in 2 dimensions
Centeres at $(0,0)$ and $(a,b)$ where $(a,b)$ is a rational point on the unit circle. 
The Picard-Fuchs operator in x1 can be read off the output for the volume. 

```python
# now move along the unit circle for 2 l2 balls in R2
#there needs to be a function the outputs just the first Pdx
def exampleonunitcircle(pt):
    n = 2
    S = PolynomialRing(QQ, "x", n)
    x = S.gens()
    mus = [[0,0], pt]

    fs = [shifted_lp_poly(S, 2, mu) for mu in mus]

    prec = 200
    strategy={"proj_var":last_variable_proj_var_strategy}
    vol = volume2(fs, prec, strategy=strategy, debug_level=1)

    print("The volume = {}".format(vol))
```

```python
#Points = [[1,0],[0,1],[-3/5,4/5],[15/17,8/17],[8/10,-6/10]]
exampleonunitcircle([0,1])
```

We can check that $P_t$ is in fact the same (up to sign) for all these points $(a,b)$.

```python
def Pt_rational_points_on_circle():
    n = 2
    S = PolynomialRing(QQ, "x", n)
    x = S.gens()
    D = {}
    for pt in [[0,1],[1,0],[3/5,4/5],[8/10,-6/10],[-15/17,-8/17]]:
        mus = [[0,0], pt]
        fs = [shifted_lp_poly(S, 2, mu) for mu in mus]
        prec = 200
        strategy={"proj_var":last_variable_proj_var_strategy}
        Pt = get_picard_fuchs_t(fs, prec, debug_level=1)
        D = D | {pt[0]:Pt}
    return D
```

```python
D = Pt_rational_points_on_circle()
D
```

```python
for key in D.keys():
    if not ((D.get(0) == D.get(key)) or (D.get(0) == -D.get(key))):
        print("Not all Picard-Fuchs operators are the same.")
        break
    
print("All Picard-Fuchs operators are the same.")
```
